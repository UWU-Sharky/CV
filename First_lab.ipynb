````python
import cv2
import numpy as np
````

#### 1. Загрузить изображение и вывести его размеры и число каналов.

````python
img = cv2.imread('Van_Gogh_-_Starry_Night_-_Google_Art_Project.jpg')
cv2.imshow('Starry_Night', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
print(f"Длина: {img.shape[1]}, Высота: {img.shape[0]} Каналы: {img.shape[2]}")
````

#### 2. Преобразовать изображение в градации серого и применить пороговую бинаризацию методом Оцу.

````python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow('dtarry_Night Gray', gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
val, otsu = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
cv2.imshow('otsu',otsu)
cv2.waitKey(0)
cv2.destroyAllWindows()
print(f"Автоматически найденный порог Оцу: {val}")
````

#### 3. Преобразовать изображение в HSV и вывести каналы H, S, V.

````python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
cv2.imshow('HSV', hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()

h, s, v = cv2.split(hsv)
cv2.imshow('Hue channel', h)
cv2.imshow('Saturation channel', s)
cv2.imshow('Value channel', v)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

#### 4. Изменить цвет всех пикселей на левом краю изображения (ширина 20 пикселей) на синий.

````python
copy = img.copy()
for h in range (copy.shape[0]):
    for w in range (copy.shape[1]):
        if w <= 20: 
            copy[h, w] = [255, 0, 0]

cv2.imshow('Starry_Night_with_blue_edge', copy)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

#### 5. Сохранить изображение с изменениями.

````python
cv2.imwrite('saved_copy.png', copy)
````

#### 6. Уменьшить изображение в 2 раза по ширине и высоте.

````python
croped_img = cv2.resize(img, (img.shape[1]//2, img.shape[0]//2), interpolation=cv2.INTER_AREA)
cv2.imshow('croped_img', croped_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

#### 7. Вывести все оттенки зеленого.

````python
image_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lower_green = np.array([35, 40, 40])
upper_green = np.array([85, 255, 255])

mask = cv2.inRange(image_hsv, lower_green, upper_green)

result = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow('result', result)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

#### 8. Нарисовать зелёную окружность с центром в середине изображения.

````python
cv2.circle(copy, (copy.shape[1]//2, copy.shape[0]//2), 200, (0,255,0), -1)
cv2.imshow('Green_Circle_in_Center', copy)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

#### 9. Отзеркалить изображение.

````python
horizontal_flip = cv2.flip(copy, 1)
cv2.imshow('horizontal_flip_img', horizontal_flip)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

#### 10. Применить сглаживающий фильтр (размытие) к изображению.

````python
blurred_img = cv2.GaussianBlur(copy, (15,15), sigmaX=0)
cv2.imshow('blurred_img', blurred_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
````

1. **Пиксель** — минимальный элемент цифрового фото. В цветном изображении он состоит из значений интенсивности базовых цветов (каналов), обычно от 0 до 255 каждый.

2. **Разница RGB и HSV:**
   * **RGB (Red, Green, Blue)** — задает цвет смешиванием красного, зеленого и синего. Удобен для мониторов, но неудобен для анализа (освещение меняет все 3 канала).
   * **HSV (Hue, Saturation, Value)** — разделяет цветовой тон (H), насыщенность (S) и яркость (V). Удобен для алгоритмов, так как цвет отделен от освещения.
